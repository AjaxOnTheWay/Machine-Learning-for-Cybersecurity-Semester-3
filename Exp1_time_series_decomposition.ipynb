```python
# =============================================================================
# Experiment 1: Time Series Decomposition
# Objective: Perform a time series decomposition of a dataset with seasonal and trend components. [CO1]
# =============================================================================

# 1. Import Required Libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from statsmodels.tsa.seasonal import seasonal_decompose

# Styling for plots
plt.style.use('seaborn-v0_8' if 'seaborn-v0_8' in plt.style.available else 'default')
import warnings
warnings.filterwarnings('ignore')
```

```python
# =============================================================================
# 2. Create / Load Time Series Data
# Generating a synthetic time series with a clear upward trend, strong seasonality, and random noise.
# =============================================================================

# Set random seed for reproducibility
np.random.seed(42)

# Create a monthly date range over 5 years (60 periods)
date_range = pd.date_range(start='2020-01-01', periods=60, freq='ME')

# Define individual components
n = len(date_range)
trend = np.linspace(50, 150, n)  # Upward linear trend
seasonality = 20 * np.sin(2 * np.pi * np.arange(n) / 12)  # Yearly seasonal cycle
noise = np.random.normal(0, 3, n)  # Random noise / residuals

# Combine components into an additive time series model: Y(t) = Trend + Seasonality + Noise
values = trend + seasonality + noise

# Create pandas Series with DatetimeIndex
ts_data = pd.Series(values, index=date_range)

# Display the first few rows of the generated time series
print("--- First 5 Rows of Time Series Data ---")
print(ts_data.head())
```

```python
# =============================================================================
# 3. Visualize the Original Time Series
# =============================================================================

plt.figure(figsize=(12, 4))
plt.plot(ts_data, label='Original Time Series', color='b', linewidth=2)
plt.title('Synthetic Time Series Data with Trend and Seasonality', fontsize=14)
plt.xlabel('Date', fontsize=12)
plt.ylabel('Value', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend()
plt.show()
```

```python
# =============================================================================
# 4. Perform Time Series Decomposition
# Using statsmodels to decompose the series into Trend, Seasonal, and Residual components.
# =============================================================================

# Perform seasonal decomposition using an additive model (period = 12 months for yearly cycle)
decomposition = seasonal_decompose(ts_data, model='additive', period=12)

# Extract individual components
trend_comp = decomposition.trend
seasonal_comp = decomposition.seasonal
residual_comp = decomposition.resid

# Combine components into a dataframe for inspection
comp_df = pd.DataFrame({
    'Original': ts_data,
    'Trend': trend_comp,
    'Seasonal': seasonal_comp,
    'Residual': residual_comp
})

print("--- Decomposition Components (First 15 Rows) ---")
print(comp_df.head(15))
```

```python
# =============================================================================
# 5. Plot Decomposed Components
# =============================================================================

fig = decomposition.plot()
fig.set_size_inches(12, 8)
plt.suptitle('Time Series Decomposition (Additive Model)', fontsize=16, y=1.02)
plt.tight_layout()
plt.show()
```
