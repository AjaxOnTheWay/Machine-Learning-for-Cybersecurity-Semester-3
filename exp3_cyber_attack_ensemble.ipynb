```python
# =============================================================================
# Experiment 3: Cyber-Attack Classification using Ensemble Learning
# Objective: Implement an ensemble learning method (e.g., Random Forest or Gradient Boosting) to classify cyber-attacks. [CO2]
# =============================================================================

# 1. Import Required Libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score

import warnings
warnings.filterwarnings('ignore')
```

```python
# =============================================================================
# 2. Generate Synthetic Network Intrusion Dataset
# Simulating network traffic features (e.g., Packet Size, Duration, Failed Logins, Byte Rate)
# Multi-class target: 0 = Normal, 1 = DoS Attack, 2 = Probe/Scan, 3 = Brute Force
# =============================================================================

X, y = make_classification(
    n_samples=2000,
    n_features=10,
    n_informative=7,
    n_redundant=2,
    n_classes=4,
    n_clusters_per_class=1,
    random_state=42
)

feature_names = [
    'Duration', 'Src_Bytes', 'Dst_Bytes', 'Packet_Rate', 'Failed_Logins',
    'Flag_Count', 'Protocol_Type', 'Service_Count', 'Error_Rate', 'Same_Srv_Rate'
]

class_names = ['Normal', 'DoS', 'Probe', 'Brute_Force']

df = pd.DataFrame(X, columns=feature_names)
df['Attack_Type'] = [class_names[i] for i in y]

print("--- Dataset Preview ---")
print(df.head())
print("\n--- Class Distribution ---")
print(df['Attack_Type'].value_counts())
```

```python
# =============================================================================
# 3. Train-Test Split
# =============================================================================

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42, stratify=y
)

print(f"Training samples: {X_train.shape[0]} | Testing samples: {X_test.shape[0]}")
```

```python
# =============================================================================
# 4. Train Ensemble Models (Random Forest & Gradient Boosting)
# =============================================================================

# Model A: Random Forest Classifier
rf_model = RandomForestClassifier(n_estimators=100, random_state=42)
rf_model.fit(X_train, y_train)
rf_preds = rf_model.predict(X_test)

# Model B: Gradient Boosting Classifier
gb_model = GradientBoostingClassifier(n_estimators=100, learning_rate=0.1, random_state=42)
gb_model.fit(X_train, y_train)
gb_preds = gb_model.predict(X_test)

print(f"Random Forest Accuracy:     {accuracy_score(y_test, rf_preds):.4f}")
print(f"Gradient Boosting Accuracy: {accuracy_score(y_test, gb_preds):.4f}")
```

```python
# =============================================================================
# 5. Evaluate Classification Performance & Feature Importance
# =============================================================================

print("--- Random Forest Classification Report ---")
print(classification_report(y_test, rf_preds, target_names=class_names))

# Plot Confusion Matrix & Feature Importance side-by-side
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Confusion Matrix (Random Forest)
cm = confusion_matrix(y_test, rf_preds)
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=class_names, yticklabels=class_names, ax=axes[0])
axes[0].set_title('Random Forest Confusion Matrix')
axes[0].set_xlabel('Predicted Class')
axes[0].set_ylabel('Actual Class')

# Feature Importances
importances = pd.Series(rf_model.feature_importances_, index=feature_names).sort_values()
importances.plot(kind='barh', color='teal', ax=axes[1])
axes[1].set_title('Feature Importance in Cyber-Attack Detection')
axes[1].set_xlabel('Importance Score')

plt.tight_layout()
plt.show()
```
